Gathering p-value and f_stat through ANOVA one-way test for each of the digital literacy categories

```python
low = df[df["digital_literacy_score"] == "Low"]["trust_score_out_of_10"]
medium = df[df["digital_literacy_score"] == "Medium"]["trust_score_out_of_10"]
high = df[df["digital_literacy_score"] == "High"]["trust_score_out_of_10"]
expert = df[df["digital_literacy_score"] == "Expert"]["trust_score_out_of_10"]

f_stat, p_value = f_oneway(low, medium, high, expert)
```

```python
print(f_stat)
```

```python
print(p_value)
```

### Turkey Test

Tukey test to gather further insight into which categories differ statistically significant

```python
turkey = pairwise_tukeyhsd(endog = df["trust_score_out_of_10"], groups = df["digital_literacy_score"],
    alpha=0.05)
```

```python
print(turkey)
```

### Feature Selection 

```python
y = df["trust_score_out_of_10"]
```

```python
dropped_col = ["trust_score_out_of_10","user_skepticism_category","trust_calibration_valid"]
```

```python
x = df.drop(columns = dropped_col)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2,random_state = 42)
```

```python
stand_scaler = StandardScaler()
```

Gathering the different data types into a list format so that specific numerical features can be scaled

```python
num_feat = x_train.select_dtypes(include=["int64","float64"]).columns.tolist()
cat_feat = x_train.select_dtypes(include=["object","str"]).columns.tolist()
bool_feat = x_train.select_dtypes(include=["bool"]).columns.tolist()
```

Scaling training and testing numerical features

```python
stand_scaler.fit(x_train[num_feat])
stand_train = stand_scaler.transform(x_train[num_feat])
stand_test = stand_scaler.transform(x_test[num_feat])
```

```python
stand_train = pd.DataFrame(stand_train, columns = num_feat, index = x_train.index)
stand_test = pd.DataFrame(stand_test, columns = num_feat, index = x_test.index)
```

Encoding the categorical and boolean features

```python
train_encode = pd.get_dummies(x_train[cat_feat], drop_first = True, dtype = int)
train_bool = x_train[bool_feat].astype(int)
```

```python
test_encode = pd.get_dummies(x_test[cat_feat], drop_first = True, dtype = int)
test_bool = x_test[bool_feat].astype(int)
```

```python
final_x_train = pd.concat([stand_train, train_encode, train_bool], axis = 1)
final_x_test = pd.concat([stand_test, test_encode, test_bool], axis = 1)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(final_x_train, y_train)
```

```python
y_pred = model.predict(final_x_test)
```

```python
y_pred
```

```python
pred_class = pd.DataFrame(y_pred)
```

```python
pred_class["Expected"] = y_test
```

```python
y_test 
```

A few model predictions agains the actual value

```python
results = pd.DataFrame({"Predicted": y_pred, "Actual":y_test})
results
```

```python
results = pd.DataFrame({"Predicted": y_pred, "Actual":y_test.reset_index(drop=True)})
results
```

Evaluation metrics used for evaluating all three models

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
print("R²:", r2)
print("MAE:", mae)
print("RMSE:", rmse)
```

```python
y_train_pred = model.predict(final_x_train)
```

```python
train_r2 = r2_score(y_train, y_train_pred)
train_mae = mean_absolute_error(y_train, y_train_pred)
train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
```

```python
print("Training R²:", train_r2)
print("Training MAE:", train_mae)
print("Training RMSE:", train_rmse)
```

### DECISITION TREE REGRESSOR

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model_tree = DecisionTreeRegressor(max_depth = 5, random_state = 42)
```

```python
model_tree.fit(final_x_train, y_train)
```

```python
train_tree_pred = model_tree.predict(final_x_train)
tree_pred = model_tree.predict(final_x_test)
```

```python
tree_r2 = r2_score(y_test, tree_pred)
tree_mae = mean_absolute_error(y_test, tree_pred)
tree_rmse = np.sqrt(mean_squared_error(y_test, tree_pred))
```

```python
print("R²:", tree_r2)
print("MAE:", tree_mae)
print("RMSE:", tree_rmse)
```

```python
print("Training R²:", r2_score(y_train, train_tree_pred))
print("Training MAE:", mean_absolute_error(y_train, train_tree_pred))
print("Training RMSE:", np.sqrt(mean_squared_error(y_train, train_tree_pred)))
```

### Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
print("R²:", rand_r2)
print("MAE:", rand_mae)
print("RMSE", rand_rmse)
```
